```python
import json
import os
import pandas as pd

folder_path = r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json"

team_name_mapping = {
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Kings XI Punjab': 'Punjab Kings',
    'Delhi Daredevils': 'Delhi Capitals',
    'Rising Pune Supergiants': 'Rising Pune Supergiant'
}

all_balls = []

for file in os.listdir(folder_path):

    if file.endswith(".json"):

        with open(os.path.join(folder_path, file), 'r') as f:
            data = json.load(f)

        match_id = file.replace('.json', '')

        for innings in data.get('innings', []):

            # Super Over exclude
            if innings.get('super_over') is True:
                continue

            batting_team = innings.get('team', '')
            batting_team = team_name_mapping.get(
                batting_team,
                batting_team
            )

            for over in innings.get('overs', []):

                over_num = over.get('over', 0)

                for delivery_number, delivery in enumerate(
                    over.get('deliveries', []),
                    start=1
                ):

                    extras = delivery.get('extras', {})

                    wides = extras.get('wides', 0)
                    noballs = extras.get('noballs', 0)
                    byes = extras.get('byes', 0)
                    legbyes = extras.get('legbyes', 0)

                    legal_ball = 0 if (wides > 0 or noballs > 0) else 1

                    # Wicket information
                    wicket = 0
                    player_out = ''
                    dismissal_kind = ''

                    if delivery.get('wickets'):

                        wicket = 1

                        first_wicket = delivery['wickets'][0]

                        player_out = first_wicket.get(
                            'player_out',
                            ''
                        )

                        dismissal_kind = first_wicket.get(
                            'kind',
                            ''
                        )

                    ball = {
                        'match_id': match_id,
                        'batting_team': batting_team,
                        'over': over_num,
                        'delivery_number': delivery_number,

                        'batter': delivery.get('batter', ''),
                        'bowler': delivery.get('bowler', ''),

                        'runs_batter': delivery['runs'].get('batter', 0),
                        'runs_extras': delivery['runs'].get('extras', 0),
                        'runs_total': delivery['runs'].get('total', 0),

                        'wides': wides,
                        'noballs': noballs,
                        'byes': byes,
                        'legbyes': legbyes,

                        'legal_ball': legal_ball,

                        'wicket': wicket,
                        'player_out': player_out,

                        # NEW
                        'dismissal_kind': dismissal_kind
                    }

                    all_balls.append(ball)

balls_df = pd.DataFrame(all_balls)

print(f"Total deliveries: {len(balls_df)}")
print("Ball-by-ball extraction completed!")

print("\nColumns:")
print(balls_df.columns.tolist())

balls_df.head()
```

```python
# ==========================================
# CORRECT RECENT TEAM BATTING STRENGTH
# Previous 5 completed batting innings only
# ==========================================

team_match_runs = (
    balls_df
    .groupby(['match_id', 'batting_team'])['runs_total']
    .sum()
    .reset_index()
)

matches_temp = pd.read_csv(
    r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json\ipl_final.csv"
)

matches_temp['match_id'] = matches_temp['match_id'].astype(str)
team_match_runs['match_id'] = team_match_runs['match_id'].astype(str)

matches_temp['date'] = pd.to_datetime(matches_temp['date'])

# Date attach karo
team_match_runs = team_match_runs.merge(
    matches_temp[['match_id', 'date']],
    on='match_id',
    how='left'
)

team_match_runs = team_match_runs.sort_values(['date', 'match_id'])

# Har team ke previous batting innings history
batting_history = {}

team1_recent_avg_runs = []
team2_recent_avg_runs = []

for _, row in matches_temp.sort_values('date').iterrows():

    match_id = row['match_id']
    team1 = row['team1']
    team2 = row['team2']

    # Current match se PEHLE ki history
    h1 = batting_history.get(team1, [])
    h2 = batting_history.get(team2, [])

    avg1 = sum(h1[-5:]) / len(h1[-5:]) if h1 else 150
    avg2 = sum(h2[-5:]) / len(h2[-5:]) if h2 else 150

    team1_recent_avg_runs.append(avg1)
    team2_recent_avg_runs.append(avg2)

    # Current match me jis team ne actually batting ki,
    # sirf uska score history me add karo
    current_scores = team_match_runs[
        team_match_runs['match_id'] == match_id
    ]

    for _, score_row in current_scores.iterrows():

        team = score_row['batting_team']
        score = score_row['runs_total']

        batting_history.setdefault(team, []).append(score)

# Features directly matches_temp me
matches_temp = matches_temp.sort_values('date').reset_index(drop=True)

matches_temp['team1_recent_avg_runs'] = team1_recent_avg_runs
matches_temp['team2_recent_avg_runs'] = team2_recent_avg_runs

print(
    matches_temp[
        [
            'date',
            'team1',
            'team2',
            'team1_recent_avg_runs',
            'team2_recent_avg_runs'
        ]
    ].tail(10)
)

print("\nMissing values:")
print(
    matches_temp[
        ['team1_recent_avg_runs', 'team2_recent_avg_runs']
    ].isnull().sum()
)
```

```python
# ==========================================
# CORRECT RECENT TEAM BOWLING STRENGTH
# Previous 5 completed matches only
# ==========================================

# Match-wise team scores
team_scores = (
    balls_df
    .groupby(['match_id', 'batting_team'])['runs_total']
    .sum()
    .reset_index()
)

team_scores['match_id'] = team_scores['match_id'].astype(str)

matches_temp = matches_temp.sort_values('date').reset_index(drop=True)

bowling_history = {}

team1_recent_runs_conceded = []
team2_recent_runs_conceded = []

for _, row in matches_temp.iterrows():

    match_id = row['match_id']
    team1 = row['team1']
    team2 = row['team2']

    # Current match se PEHLE ki bowling history
    h1 = bowling_history.get(team1, [])
    h2 = bowling_history.get(team2, [])

    avg1 = sum(h1[-5:]) / len(h1[-5:]) if h1 else 150
    avg2 = sum(h2[-5:]) / len(h2[-5:]) if h2 else 150

    team1_recent_runs_conceded.append(avg1)
    team2_recent_runs_conceded.append(avg2)

    # Current match scores
    current_scores = team_scores[
        team_scores['match_id'] == match_id
    ]

    team1_score = current_scores.loc[
        current_scores['batting_team'] == team1,
        'runs_total'
    ]

    team2_score = current_scores.loc[
        current_scores['batting_team'] == team2,
        'runs_total'
    ]

    # Team1 conceded Team2's score
    if len(team2_score) > 0:
        bowling_history.setdefault(team1, []).append(
            team2_score.iloc[0]
        )

    # Team2 conceded Team1's score
    if len(team1_score) > 0:
        bowling_history.setdefault(team2, []).append(
            team1_score.iloc[0]
        )

matches_temp['team1_recent_runs_conceded'] = team1_recent_runs_conceded
matches_temp['team2_recent_runs_conceded'] = team2_recent_runs_conceded

print(
    matches_temp[
        [
            'date',
            'team1',
            'team2',
            'team1_recent_runs_conceded',
            'team2_recent_runs_conceded'
        ]
    ].tail(10)
)

print("\nMissing values:")
print(
    matches_temp[
        [
            'team1_recent_runs_conceded',
            'team2_recent_runs_conceded'
        ]
    ].isnull().sum()
)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

# Final feature set = old 9 + new 4 ball-based features
features_ball = [
    'team1_encoded',
    'team2_encoded',
    'venue_encoded',
    'toss_winner_encoded',
    'bat_first',
    'season',
    'h2h_win_rate',
    'team1_recent_form',
    'team2_recent_form',
    'team1_recent_avg_runs',
    'team2_recent_avg_runs',
    'team1_recent_runs_conceded',
    'team2_recent_runs_conceded'
]

X_ball = matches_temp[features_ball]
Y_ball = matches_temp['winner_encoded']

X_train_ball, X_test_ball, Y_train_ball, Y_test_ball = train_test_split(
    X_ball,
    Y_ball,
    test_size=0.2,
    random_state=42
)

gradient_ball_model = GradientBoostingClassifier(
    n_estimators=200,
    random_state=42
)

gradient_ball_model.fit(
    X_train_ball,
    Y_train_ball
)

Y_pred_ball = gradient_ball_model.predict(
    X_test_ball
)

ball_accuracy = accuracy_score(
    Y_test_ball,
    Y_pred_ball
)

print("Previous Gradient Boosting: 54.22%")
print(f"Gradient Boosting + Ball Features: {ball_accuracy * 100:.2f}%")
```

```python
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    HistGradientBoostingClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import accuracy_score
import pandas as pd


# ==========================================
# MODEL COMPARISON
# Same train/test data for every model
# ==========================================

models = {
    'Random Forest': RandomForestClassifier(
        n_estimators=300,
        random_state=42
    ),

    'Gradient Boosting': GradientBoostingClassifier(
        n_estimators=200,
        random_state=42
    ),

    'Hist Gradient Boosting': HistGradientBoostingClassifier(
        max_iter=200,
        random_state=42
    ),

    'Extra Trees': ExtraTreesClassifier(
        n_estimators=300,
        random_state=42
    )
}


results = []

for name, current_model in models.items():

    current_model.fit(X_train, Y_train)

    predictions = current_model.predict(X_test)

    accuracy = accuracy_score(
        Y_test,
        predictions
    )

    results.append({
        'Model': name,
        'Accuracy': round(accuracy * 100, 2)
    })


# Results dataframe
model_results = pd.DataFrame(results)

model_results = model_results.sort_values(
    by='Accuracy',
    ascending=False
).reset_index(drop=True)

print(model_results.to_string(index=False))
```

```python
import pandas as pd

# ==========================================
# FINAL BATTING STATS
# ==========================================

# Legal balls faced:
# wides batter ki ball faced me count nahi hote
batting_stats = balls_df.groupby('batter').agg(
    total_runs=('runs_batter', 'sum')
).reset_index()

balls_faced = (
    balls_df[balls_df['wides'] == 0]
    .groupby('batter')
    .size()
    .reset_index(name='total_balls')
)

batting_stats = batting_stats.merge(
    balls_faced,
    on='batter',
    how='left'
)

# Dismissal types that count against batter
valid_dismissals = [
    'bowled',
    'caught',
    'caught and bowled',
    'hit wicket',
    'lbw',
    'obstructing the field',
    'run out',
    'stumped',
    'retired out'
]

dismissals = (
    balls_df[
        (balls_df['player_out'] != '') &
        (balls_df['player_out'].notna())
    ]
    .groupby('player_out')
    .size()
    .reset_index(name='dismissals')
)

dismissals = dismissals.rename(
    columns={'player_out': 'batter'}
)

batting_stats = batting_stats.merge(
    dismissals,
    on='batter',
    how='left'
)

batting_stats['dismissals'] = (
    batting_stats['dismissals']
    .fillna(0)
    .astype(int)
)

batting_stats['total_balls'] = (
    batting_stats['total_balls']
    .fillna(0)
    .astype(int)
)

# Strike Rate
batting_stats['strike_rate'] = (
    batting_stats['total_runs'] /
    batting_stats['total_balls'] * 100
).round(2)

# Batting Average
batting_stats['average'] = (
    batting_stats['total_runs'] /
    batting_stats['dismissals'].replace(0, float('nan'))
).round(2)


# ==========================================
# FINAL BOWLING STATS
# ==========================================

# Runs conceded by bowler:
# byes + legbyes bowler ke against count nahi hote
balls_df['bowler_runs'] = (
    balls_df['runs_total']
    - balls_df['byes']
    - balls_df['legbyes']
)

# IMPORTANT:
# 'wicket' column alone use nahi karenge,
# kyunki run out etc bowler wicket nahi hote.

bowler_wicket_types = [
    'bowled',
    'caught',
    'caught and bowled',
    'hit wicket',
    'lbw',
    'stumped'
]

# Agar dismissal_kind extraction me available nahi hai,
# pehle isko verify karna hoga.
if 'dismissal_kind' not in balls_df.columns:
    print("⚠️ dismissal_kind column missing hai.")
    print("Bowling wickets abhi calculate/save MAT karna.")
else:

    balls_df['bowler_wicket'] = (
        balls_df['dismissal_kind']
        .isin(bowler_wicket_types)
        .astype(int)
    )

    bowling_stats = balls_df.groupby('bowler').agg(
        total_balls=('legal_ball', 'sum'),
        total_runs=('bowler_runs', 'sum'),
        total_wickets=('bowler_wicket', 'sum')
    ).reset_index()

    # Proper cricket overs format
    bowling_stats['overs'] = (
        (bowling_stats['total_balls'] // 6).astype(str)
        + '.'
        + (bowling_stats['total_balls'] % 6).astype(str)
    )

    # Economy = runs per 6 legal balls
    bowling_stats['economy'] = (
        bowling_stats['total_runs'] * 6 /
        bowling_stats['total_balls'].replace(0, float('nan'))
    ).round(2)

    bowling_stats['bowling_avg'] = (
        bowling_stats['total_runs'] /
        bowling_stats['total_wickets'].replace(0, float('nan'))
    ).round(2)

print("\nTop Batters:")
print(
    batting_stats
    .sort_values('total_runs', ascending=False)
    .head(10)
    .to_string(index=False)
)

if 'bowling_stats' in globals():
    print("\nTop Bowlers:")
    print(
        bowling_stats
        .sort_values('total_wickets', ascending=False)
        .head(10)
        .to_string(index=False)
    )
```

```python
# ==========================================
# FINAL CORRECT BOWLING STATS
# ==========================================

# Bowler ko credit milne wale wicket types
import numpy as np
bowler_wicket_kinds = [
    'bowled',
    'caught',
    'caught and bowled',
    'hit wicket',
    'lbw',
    'stumped'
]

# Legal balls + bowler conceded runs + wickets collect karo
bowling_records = []

for file in os.listdir(folder_path):

    if file.endswith(".json"):

        with open(os.path.join(folder_path, file), 'r') as f:
            data = json.load(f)

        for innings in data.get('innings', []):

            # Super Over exclude
            if innings.get('super_over') is True:
                continue

            for over in innings.get('overs', []):

                for delivery in over.get('deliveries', []):

                    bowler = delivery.get('bowler', '')

                    extras = delivery.get('extras', {})

                    wides = extras.get('wides', 0)
                    noballs = extras.get('noballs', 0)
                    byes = extras.get('byes', 0)
                    legbyes = extras.get('legbyes', 0)

                    # Legal ball: wides/no-balls count nahi honge
                    legal_ball = 0 if (wides > 0 or noballs > 0) else 1

                    # Bowler conceded runs:
                    # byes + leg byes bowler ke against count nahi hote
                    total_runs = delivery['runs'].get('total', 0)
                    bowler_runs = total_runs - byes - legbyes

                    # Bowler wickets
                    bowler_wickets = 0

                    for wicket in delivery.get('wickets', []):
                        if wicket.get('kind') in bowler_wicket_kinds:
                            bowler_wickets += 1

                    bowling_records.append({
                        'bowler': bowler,
                        'legal_ball': legal_ball,
                        'bowler_runs': bowler_runs,
                        'bowler_wickets': bowler_wickets
                    })

bowling_df = pd.DataFrame(bowling_records)

bowling_stats = bowling_df.groupby('bowler').agg(
    total_balls=('legal_ball', 'sum'),
    total_runs=('bowler_runs', 'sum'),
    total_wickets=('bowler_wickets', 'sum')
).reset_index()

# Overs proper format me
bowling_stats['overs'] = (
    (bowling_stats['total_balls'] // 6).astype(str)
    + '.'
    + (bowling_stats['total_balls'] % 6).astype(str)
)

# Economy
bowling_stats['economy'] = (
    bowling_stats['total_runs'] /
    (bowling_stats['total_balls'] / 6)
).round(2)

# Bowling Average
bowling_stats['bowling_avg'] = (
    bowling_stats['total_runs'] /
    bowling_stats['total_wickets'].replace(0, np.nan)
).round(2)

print(
    bowling_stats
    .sort_values('total_wickets', ascending=False)
    .head(15)
    .to_string(index=False)
)
```

```python
# Recent Team Batting Strength (previous 5 matches)

# Match-wise team total score
team_match_runs = (
    balls_df
    .groupby(['match_id', 'batting_team'])['runs_total']
    .sum()
    .reset_index()
)

# Match date add karo
matches_temp = pd.read_csv(
    r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json\ipl_final.csv"
)

matches_temp['match_id'] = matches_temp['match_id'].astype(str)
team_match_runs['match_id'] = team_match_runs['match_id'].astype(str)

team_match_runs = team_match_runs.merge(
    matches_temp[['match_id', 'date']],
    on='match_id',
    how='left'
)

team_match_runs['date'] = pd.to_datetime(team_match_runs['date'])

team_match_runs = team_match_runs.sort_values('date')

# Previous 5 match batting average
team_match_runs['recent_avg_runs'] = (
    team_match_runs
    .groupby('batting_team')['runs_total']
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=1).mean()
    )
)

team_match_runs['recent_avg_runs'] = (
    team_match_runs['recent_avg_runs'].fillna(150)
)

print(
    team_match_runs[
        ['date', 'batting_team', 'runs_total', 'recent_avg_runs']
    ].tail(20)
)
```

```python
# Recent Team Bowling Strength
# Previous 5 matches me opponent ko kitne runs concede kiye

# Match-wise innings totals
team_scores = (
    balls_df
    .groupby(['match_id', 'batting_team'])['runs_total']
    .sum()
    .reset_index()
)

matches_temp = pd.read_csv(
    r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json\ipl_final.csv"
)

matches_temp['match_id'] = matches_temp['match_id'].astype(str)
team_scores['match_id'] = team_scores['match_id'].astype(str)

# Team1 / Team2 ke opponent runs nikalo
bowling_rows = []

for _, row in matches_temp.iterrows():
    match_id = row['match_id']
    team1 = row['team1']
    team2 = row['team2']
    date = row['date']

    scores = team_scores[team_scores['match_id'] == match_id]

    team1_conceded = scores.loc[
        scores['batting_team'] == team2,
        'runs_total'
    ].sum()

    team2_conceded = scores.loc[
        scores['batting_team'] == team1,
        'runs_total'
    ].sum()

    bowling_rows.append({
        'match_id': match_id,
        'date': date,
        'team': team1,
        'runs_conceded': team1_conceded
    })

    bowling_rows.append({
        'match_id': match_id,
        'date': date,
        'team': team2,
        'runs_conceded': team2_conceded
    })

team_bowling = pd.DataFrame(bowling_rows)

team_bowling['date'] = pd.to_datetime(team_bowling['date'])
team_bowling = team_bowling.sort_values('date')

# Previous 5 matches average runs conceded
team_bowling['recent_avg_runs_conceded'] = (
    team_bowling
    .groupby('team')['runs_conceded']
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=1).mean()
    )
)

team_bowling['recent_avg_runs_conceded'] = (
    team_bowling['recent_avg_runs_conceded']
    .fillna(150)
)

print(
    team_bowling[
        ['date', 'team', 'runs_conceded', 'recent_avg_runs_conceded']
    ].tail(20)
)
```

```python
# ==========================================
# MERGE RECENT BATTING + BOWLING FEATURES
# ==========================================

matches_model = matches_temp.copy()

# Ensure same type
matches_model['match_id'] = matches_model['match_id'].astype(str)
team_match_runs['match_id'] = team_match_runs['match_id'].astype(str)
team_bowling['match_id'] = team_bowling['match_id'].astype(str)

# -----------------------------
# TEAM 1 BATTING FEATURE
# -----------------------------
team1_batting = team_match_runs[
    ['match_id', 'batting_team', 'recent_avg_runs']
].rename(columns={
    'batting_team': 'team1',
    'recent_avg_runs': 'team1_recent_avg_runs'
})

matches_model = matches_model.merge(
    team1_batting,
    on=['match_id', 'team1'],
    how='left'
)

# -----------------------------
# TEAM 2 BATTING FEATURE
# -----------------------------
team2_batting = team_match_runs[
    ['match_id', 'batting_team', 'recent_avg_runs']
].rename(columns={
    'batting_team': 'team2',
    'recent_avg_runs': 'team2_recent_avg_runs'
})

matches_model = matches_model.merge(
    team2_batting,
    on=['match_id', 'team2'],
    how='left'
)

# -----------------------------
# TEAM 1 BOWLING FEATURE
# -----------------------------
team1_bowling = team_bowling[
    ['match_id', 'team', 'recent_avg_runs_conceded']
].rename(columns={
    'team': 'team1',
    'recent_avg_runs_conceded': 'team1_recent_runs_conceded'
})

matches_model = matches_model.merge(
    team1_bowling,
    on=['match_id', 'team1'],
    how='left'
)

# -----------------------------
# TEAM 2 BOWLING FEATURE
# -----------------------------
team2_bowling = team_bowling[
    ['match_id', 'team', 'recent_avg_runs_conceded']
].rename(columns={
    'team': 'team2',
    'recent_avg_runs_conceded': 'team2_recent_runs_conceded'
})

matches_model = matches_model.merge(
    team2_bowling,
    on=['match_id', 'team2'],
    how='left'
)

print(
    matches_model[
        [
            'date',
            'team1',
            'team2',
            'team1_recent_avg_runs',
            'team2_recent_avg_runs',
            'team1_recent_runs_conceded',
            'team2_recent_runs_conceded'
        ]
    ].tail(10)
)
```

```python
import pandas as pd

# ==========================================
# FINAL CORRECT BOWLING STATS
# ==========================================

# Bowler ke against byes aur leg byes count nahi hote
balls_df['bowler_runs'] = (
    balls_df['runs_total']
    - balls_df['byes']
    - balls_df['legbyes']
)

# Sirf ye wickets bowler ko milti hain
bowler_wicket_types = [
    'bowled',
    'caught',
    'caught and bowled',
    'hit wicket',
    'lbw',
    'stumped'
]

balls_df['bowler_wicket'] = (
    balls_df['dismissal_kind']
    .isin(bowler_wicket_types)
    .astype(int)
)

# Bowling stats
bowling_stats = balls_df.groupby('bowler').agg(
    total_balls=('legal_ball', 'sum'),
    total_runs=('bowler_runs', 'sum'),
    total_wickets=('bowler_wicket', 'sum')
).reset_index()

# Proper overs format (e.g. 471.1)
bowling_stats['overs'] = (
    (bowling_stats['total_balls'] // 6).astype(str)
    + '.'
    + (bowling_stats['total_balls'] % 6).astype(str)
)

# Economy
bowling_stats['economy'] = (
    bowling_stats['total_runs'] * 6 /
    bowling_stats['total_balls'].replace(0, float('nan'))
).round(2)

# Bowling average
bowling_stats['bowling_avg'] = (
    bowling_stats['total_runs'] /
    bowling_stats['total_wickets'].replace(0, float('nan'))
).round(2)

print(
    bowling_stats
    .sort_values('total_wickets', ascending=False)
    .head(15)
    .to_string(index=False)
)
```

```python
# ==========================================
# CREATE IPL VECTOR DATABASE USING CHROMADB
# ==========================================

import chromadb
from chromadb.utils import embedding_functions
import os
import pandas as pd

folder_path = r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json"

rag_path = os.path.join(
    folder_path,
    "ipl_rag_knowledge.csv"
)

rag_df = pd.read_csv(rag_path)

# ------------------------------------------
# Persistent Chroma Database
# ------------------------------------------

chroma_path = os.path.join(
    folder_path,
    "ipl_chroma_db"
)

client = chromadb.PersistentClient(
    path=chroma_path
)

# ------------------------------------------
# Embedding Model
# ------------------------------------------

embedding_function = (
    embedding_functions
    .SentenceTransformerEmbeddingFunction(
        model_name="all-MiniLM-L6-v2"
    )
)

# ------------------------------------------
# Delete old collection if it already exists
# ------------------------------------------

try:
    client.delete_collection(
        name="ipl_knowledge"
    )
except:
    pass


# ------------------------------------------
# Create Collection
# ------------------------------------------

collection = client.create_collection(
    name="ipl_knowledge",
    embedding_function=embedding_function
)


# ------------------------------------------
# Prepare Documents
# ------------------------------------------

documents = rag_df['text'].astype(str).tolist()

ids = [
    f"ipl_{i}"
    for i in range(len(rag_df))
]

metadatas = []

for _, row in rag_df.iterrows():

    metadatas.append({
        "type": str(row['type']),
        "entity": str(row['entity'])
    })


# ------------------------------------------
# Insert Into Vector Database
# ------------------------------------------

batch_size = 500

for start in range(
    0,
    len(documents),
    batch_size
):

    end = min(
        start + batch_size,
        len(documents)
    )

    collection.add(
        documents=documents[start:end],
        ids=ids[start:end],
        metadatas=metadatas[start:end]
    )

    print(
        f"Inserted {end}/{len(documents)} documents"
    )


print("\n✅ IPL Chroma Vector Database Created!")
print(
    "Total documents:",
    collection.count()
)

print(
    "Database location:",
    chroma_path
)
```

```python
# ============================================================
# IPL ADVANCED SEASON KNOWLEDGE
# Players, POTM, Orange Cap, Purple Cap, Champions, Titles
# ============================================================

import os
import json
import pandas as pd

folder_path = r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json"


# ============================================================
# TEAM NAME NORMALIZATION
# ============================================================

team_name_mapping = {
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Kings XI Punjab': 'Punjab Kings',
    'Delhi Daredevils': 'Delhi Capitals',
    'Rising Pune Supergiants': 'Rising Pune Supergiant'
}


def normalize_team(team):
    return team_name_mapping.get(team, team)


# ============================================================
# LOAD MATCH DATA
# ============================================================

matches = pd.read_csv(
    os.path.join(folder_path, "ipl_final.csv")
)

matches['match_id'] = matches['match_id'].astype(str)
matches['date'] = pd.to_datetime(matches['date'])

if 'season' not in matches.columns:
    matches['season'] = matches['date'].dt.year


for col in ['team1', 'team2', 'toss_winner', 'winner']:
    if col in matches.columns:
        matches[col] = matches[col].replace(team_name_mapping)


print("✅ Match data loaded")


# ============================================================
# 1. SEASON-WISE TEAM PLAYERS
# ============================================================

season_player_records = []


for file in os.listdir(folder_path):

    if not file.endswith(".json"):
        continue

    file_path = os.path.join(folder_path, file)

    with open(file_path, "r") as f:
        data = json.load(f)

    info = data.get("info", {})

    dates = info.get("dates", [])

    if not dates:
        continue

    season = pd.to_datetime(dates[0]).year

    players_data = info.get("players", {})

    for team, players in players_data.items():

        team = normalize_team(team)

        for player in players:

            season_player_records.append({
                "season": season,
                "team": team,
                "player": player
            })


season_players = pd.DataFrame(
    season_player_records
).drop_duplicates()


season_players = season_players.sort_values(
    ['season', 'team', 'player']
).reset_index(drop=True)


season_players.to_csv(
    os.path.join(
        folder_path,
        "season_team_players.csv"
    ),
    index=False
)


print(
    "✅ Season-wise player database created:",
    len(season_players)
)


# ============================================================
# 2. PLAYER OF THE MATCH DATA
# ============================================================

potm_columns = [
    'match_id',
    'season',
    'date',
    'team1',
    'team2',
    'venue',
    'winner',
    'player_of_match'
]


potm_matches = matches[
    [
        col
        for col in potm_columns
        if col in matches.columns
    ]
].copy()


potm_matches.to_csv(
    os.path.join(
        folder_path,
        "player_of_match.csv"
    ),
    index=False
)


# ============================================================
# POTM TOTAL AWARDS
# ============================================================

valid_potm = potm_matches[
    potm_matches['player_of_match'].notna()
].copy()


valid_potm = valid_potm[
    valid_potm['player_of_match'] != 'Unknown'
]


potm_summary = (
    valid_potm
    .groupby('player_of_match')
    .size()
    .reset_index(name='potm_awards')
    .sort_values(
        'potm_awards',
        ascending=False
    )
    .reset_index(drop=True)
)


potm_summary.to_csv(
    os.path.join(
        folder_path,
        "potm_summary.csv"
    ),
    index=False
)


print(
    "✅ Player of the Match database created"
)


# ============================================================
# 3. LOAD BALL DATA FOR ORANGE / PURPLE CAP
# ============================================================

balls = pd.read_csv(
    os.path.join(
        folder_path,
        "ipl_balls.csv"
    )
)


balls['match_id'] = balls['match_id'].astype(str)


# ============================================================
# ADD SEASON TO BALL DATA
# ============================================================

season_map = matches[
    ['match_id', 'season']
].drop_duplicates()


balls = balls.merge(
    season_map,
    on='match_id',
    how='left'
)


print("✅ Season added to ball-by-ball data")


# ============================================================
# 4. ORANGE CAP — MOST RUNS EACH SEASON
# ============================================================

season_batting = (
    balls
    .groupby(
        ['season', 'batter']
    )['runs_batter']
    .sum()
    .reset_index(
        name='runs'
    )
)


season_batting['rank'] = (
    season_batting
    .groupby('season')['runs']
    .rank(
        method='min',
        ascending=False
    )
)


orange_cap = (
    season_batting[
        season_batting['rank'] == 1
    ]
    .drop(columns='rank')
    .sort_values('season')
    .reset_index(drop=True)
)


orange_cap.to_csv(
    os.path.join(
        folder_path,
        "orange_cap.csv"
    ),
    index=False
)


print("✅ Orange Cap data created")


# ============================================================
# 5. PURPLE CAP — MOST WICKETS EACH SEASON
# ============================================================

bowler_wicket_types = [
    'bowled',
    'caught',
    'caught and bowled',
    'hit wicket',
    'lbw',
    'stumped'
]


# Agar bowler_wicket column saved nahi hai
if 'bowler_wicket' not in balls.columns:

    if 'dismissal_kind' in balls.columns:

        balls['bowler_wicket'] = (
            balls['dismissal_kind']
            .isin(bowler_wicket_types)
            .astype(int)
        )

    else:

        raise ValueError(
            "dismissal_kind column missing from ipl_balls.csv"
        )


season_bowling = (
    balls
    .groupby(
        ['season', 'bowler']
    )['bowler_wicket']
    .sum()
    .reset_index(
        name='wickets'
    )
)


season_bowling['rank'] = (
    season_bowling
    .groupby('season')['wickets']
    .rank(
        method='min',
        ascending=False
    )
)


purple_cap = (
    season_bowling[
        season_bowling['rank'] == 1
    ]
    .drop(columns='rank')
    .sort_values('season')
    .reset_index(drop=True)
)


purple_cap.to_csv(
    os.path.join(
        folder_path,
        "purple_cap.csv"
    ),
    index=False
)


print("✅ Purple Cap data created")


# ============================================================
# 6. IPL CHAMPIONS / FINAL WINNERS
# ============================================================

final_records = []


for file in os.listdir(folder_path):

    if not file.endswith(".json"):
        continue

    file_path = os.path.join(
        folder_path,
        file
    )

    with open(
        file_path,
        "r"
    ) as f:

        data = json.load(f)


    info = data.get(
        "info",
        {}
    )


    dates = info.get(
        "dates",
        []
    )


    if not dates:
        continue


    season = pd.to_datetime(
        dates[0]
    ).year


    event = info.get(
        "event",
        {}
    )


    stage = str(
        event.get(
            "stage",
            ""
        )
    ).strip()


    # Exact IPL Final
    if stage.lower() == "final":

        winner = (
            info
            .get(
                "outcome",
                {}
            )
            .get(
                "winner"
            )
        )


        if winner:

            winner = normalize_team(
                winner
            )


            final_records.append({
                "season": season,
                "champion": winner,
                "match_id": file.replace(
                    ".json",
                    ""
                )
            })


champions = pd.DataFrame(
    final_records
)


# ============================================================
# FALLBACK IF FINAL STAGE MISSING
# ============================================================

found_seasons = (
    set(champions['season'])
    if len(champions) > 0
    else set()
)


all_seasons = set(
    matches['season'].unique()
)


missing_final_seasons = (
    all_seasons
    - found_seasons
)


fallback_records = []


for season in missing_final_seasons:

    season_matches = (
        matches[
            matches['season'] == season
        ]
        .sort_values('date')
    )


    if len(season_matches) == 0:
        continue


    last_match = (
        season_matches.iloc[-1]
    )


    winner = last_match['winner']


    if (
        pd.notna(winner)
        and winner != 'No Result'
    ):

        fallback_records.append({
            "season": season,
            "champion": winner,
            "match_id": str(
                last_match['match_id']
            )
        })


if fallback_records:

    fallback_df = pd.DataFrame(
        fallback_records
    )

    champions = pd.concat(
        [
            champions,
            fallback_df
        ],
        ignore_index=True
    )


champions = (
    champions
    .drop_duplicates(
        subset='season'
    )
    .sort_values('season')
    .reset_index(drop=True)
)


champions.to_csv(
    os.path.join(
        folder_path,
        "ipl_champions.csv"
    ),
    index=False
)


print("✅ IPL Champions data created")


# ============================================================
# 7. TOTAL IPL TITLES PER TEAM
# ============================================================

title_counts = (
    champions
    .groupby('champion')
    .size()
    .reset_index(
        name='titles'
    )
    .sort_values(
        'titles',
        ascending=False
    )
    .reset_index(drop=True)
)


title_counts.to_csv(
    os.path.join(
        folder_path,
        "team_title_counts.csv"
    ),
    index=False
)


print("✅ Team title counts created")


# ============================================================
# 8. SEASON SUMMARY
# ============================================================

orange_summary = (
    orange_cap
    .groupby('season')
    .agg({
        'batter':
            lambda x: ', '.join(
                x.astype(str)
            ),

        'runs':
            'max'
    })
    .reset_index()
    .rename(columns={
        'batter': 'orange_cap',
        'runs': 'orange_cap_runs'
    })
)


purple_summary = (
    purple_cap
    .groupby('season')
    .agg({
        'bowler':
            lambda x: ', '.join(
                x.astype(str)
            ),

        'wickets':
            'max'
    })
    .reset_index()
    .rename(columns={
        'bowler': 'purple_cap',
        'wickets': 'purple_cap_wickets'
    })
)


season_summary = champions[
    [
        'season',
        'champion'
    ]
].copy()


season_summary = season_summary.merge(
    orange_summary,
    on='season',
    how='left'
)


season_summary = season_summary.merge(
    purple_summary,
    on='season',
    how='left'
)


season_summary.to_csv(
    os.path.join(
        folder_path,
        "season_summary.csv"
    ),
    index=False
)


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 60)
print("🔥 IPL ADVANCED DATABASE COMPLETE")
print("=" * 60)


print("\n🏆 IPL Champions:")
print(
    champions.to_string(
        index=False
    )
)


print("\n🏆 Most Successful Teams:")
print(
    title_counts.to_string(
        index=False
    )
)


print("\n🟠 Orange Cap Winners:")
print(
    orange_cap.tail(10).to_string(
        index=False
    )
)


print("\n🟣 Purple Cap Winners:")
print(
    purple_cap.tail(10).to_string(
        index=False
    )
)


print("\n⭐ Most Player of the Match Awards:")
print(
    potm_summary.head(10).to_string(
        index=False
    )
)


print("\n👥 Season-Team-Player rows:")
print(
    len(season_players)
)


print("\n✅ Files Created:")

print("season_team_players.csv")
print("player_of_match.csv")
print("potm_summary.csv")
print("orange_cap.csv")
print("purple_cap.csv")
print("ipl_champions.csv")
print("team_title_counts.csv")
print("season_summary.csv")
```

```python
# ============================================================
# CREATE FINAL IPL CHROMA VECTOR DATABASE
# ============================================================

import os
import pandas as pd
import chromadb

from chromadb.utils import embedding_functions


# ============================================================
# PATHS
# ============================================================

folder_path = r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json"

rag_file = os.path.join(
    folder_path,
    "ipl_rag_knowledge_final.csv"
)

chroma_path = os.path.join(
    folder_path,
    "ipl_chroma_db"
)


# ============================================================
# LOAD FINAL RAG KNOWLEDGE
# ============================================================

rag_df = pd.read_csv(rag_file)

print("Knowledge documents:", len(rag_df))


# ============================================================
# EMBEDDING MODEL
# ============================================================

embedding_function = (
    embedding_functions
    .SentenceTransformerEmbeddingFunction(
        model_name="all-MiniLM-L6-v2"
    )
)


# ============================================================
# CHROMA CLIENT
# ============================================================

client = chromadb.PersistentClient(
    path=chroma_path
)


# ============================================================
# DELETE OLD FINAL COLLECTION IF IT EXISTS
# ============================================================

collection_name = "ipl_knowledge_final"

try:
    client.delete_collection(
        name=collection_name
    )

    print("🗑️ Old vector collection removed")

except Exception:
    print("ℹ️ No old final collection found")


# ============================================================
# CREATE NEW COLLECTION
# ============================================================

collection = client.create_collection(
    name=collection_name,
    embedding_function=embedding_function
)


# ============================================================
# PREPARE DATA
# ============================================================

documents = (
    rag_df["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

ids = [
    f"ipl_doc_{i}"
    for i in range(len(rag_df))
]

metadatas = []

for _, row in rag_df.iterrows():

    metadatas.append({
        "type": str(row["type"]),
        "entity": str(row["entity"]),
        "season": str(row["season"])
    })


# ============================================================
# ADD DOCUMENTS IN BATCHES
# ============================================================

batch_size = 500

for start in range(
    0,
    len(documents),
    batch_size
):

    end = min(
        start + batch_size,
        len(documents)
    )

    collection.add(
        ids=ids[start:end],
        documents=documents[start:end],
        metadatas=metadatas[start:end]
    )

    print(
        f"✅ Inserted {end}/{len(documents)} documents"
    )


# ============================================================
# FINAL CHECK
# ============================================================

print("\n" + "=" * 55)
print("🔥 FINAL IPL VECTOR DATABASE READY")
print("=" * 55)

print(
    "Total vectors:",
    collection.count()
)

print(
    "Database folder:",
    chroma_path
)

print(
    "Collection:",
    collection_name
)
```

```python
# ============================================================
# SMART IPL RAG RETRIEVER
# Metadata filtering + semantic search
# ============================================================

import re
import pandas as pd
import os

folder_path = r"C:\Users\dell\OneDrive\Desktop\New Folder\Projects using DS\ipl_json"

rag_df = pd.read_csv(
    os.path.join(folder_path, "ipl_rag_knowledge_final.csv")
)

# ------------------------------------------------------------
# TEAM ALIASES
# ------------------------------------------------------------

team_aliases = {
    "rcb": "Royal Challengers Bengaluru",
    "royal challengers": "Royal Challengers Bengaluru",
    "royal challengers bangalore": "Royal Challengers Bengaluru",
    "royal challengers bengaluru": "Royal Challengers Bengaluru",

    "mi": "Mumbai Indians",
    "mumbai": "Mumbai Indians",
    "mumbai indians": "Mumbai Indians",

    "csk": "Chennai Super Kings",
    "chennai": "Chennai Super Kings",
    "chennai super kings": "Chennai Super Kings",

    "kkr": "Kolkata Knight Riders",
    "kolkata": "Kolkata Knight Riders",
    "kolkata knight riders": "Kolkata Knight Riders",

    "srh": "Sunrisers Hyderabad",
    "sunrisers": "Sunrisers Hyderabad",
    "sunrisers hyderabad": "Sunrisers Hyderabad",

    "rr": "Rajasthan Royals",
    "rajasthan": "Rajasthan Royals",
    "rajasthan royals": "Rajasthan Royals",

    "pbks": "Punjab Kings",
    "punjab kings": "Punjab Kings",
    "kings xi punjab": "Punjab Kings",

    "dc": "Delhi Capitals",
    "delhi capitals": "Delhi Capitals",
    "delhi daredevils": "Delhi Capitals",

    "gt": "Gujarat Titans",
    "gujarat titans": "Gujarat Titans",

    "lsg": "Lucknow Super Giants",
    "lucknow super giants": "Lucknow Super Giants"
}


# ------------------------------------------------------------
# FIND SEASON FROM QUESTION
# ------------------------------------------------------------

def detect_season(question):
    match = re.search(r"\b(20\d{2})\b", question)

    if match:
        return match.group(1)

    return None


# ------------------------------------------------------------
# FIND TEAM FROM QUESTION
# ------------------------------------------------------------

def detect_team(question):

    q = question.lower()

    # longer names first
    aliases = sorted(
        team_aliases.keys(),
        key=len,
        reverse=True
    )

    for alias in aliases:

        pattern = r"\b" + re.escape(alias) + r"\b"

        if re.search(pattern, q):
            return team_aliases[alias]

    return None


# ------------------------------------------------------------
# QUESTION TYPE DETECTION
# ------------------------------------------------------------

def detect_question_type(question):

    q = question.lower()

    # Squad / players
    if (
        "squad" in q
        or "players were" in q
        or "players in" in q
        or "team players" in q
        or "which players" in q
    ):
        return "season_squad"

    # Orange Cap
    if "orange cap" in q:
        return "season_summary"

    # Purple Cap
    if "purple cap" in q:
        return "season_summary"

    # Champion
    if (
        "champion" in q
        or "won the ipl" in q
        or "ipl winner" in q
        or "who won ipl" in q
    ):
        return "season_summary"

    # Team titles
    if (
        "titles" in q
        or "title" in q
        or "trophies" in q
        or "trophy" in q
    ):
        return "team_titles"

    # Player of Match totals
    if (
        "player of the match awards" in q
        or "potm awards" in q
        or "man of the match awards" in q
    ):
        return "potm_total"

    # Specific match POTM
    if (
        "player of the match" in q
        or "man of the match" in q
        or "potm" in q
    ):
        return "match"

    # Bowling
    if (
        "bowling" in q
        or "wickets" in q
        or "economy" in q
        or "bowler" in q
    ):
        return "career_bowling"

    # Batting
    if (
        "batting" in q
        or "runs" in q
        or "strike rate" in q
        or "average" in q
        or "batsman" in q
        or "batter" in q
    ):
        return "career_batting"

    return None


# ------------------------------------------------------------
# SMART RETRIEVAL FUNCTION
# ------------------------------------------------------------

def smart_retrieve(question, n_results=3):

    season = detect_season(question)
    team = detect_team(question)
    qtype = detect_question_type(question)

    filters = []

    if qtype:
        filters.append({
            "type": qtype
        })

    if season:
        filters.append({
            "season": season
        })

    # Exact team metadata filtering only for categories
    # where entity itself is the team
    if team and qtype in [
        "season_squad",
        "team_titles"
    ]:
        filters.append({
            "entity": team
        })

    # Build Chroma WHERE
    where_filter = None

    if len(filters) == 1:
        where_filter = filters[0]

    elif len(filters) > 1:
        where_filter = {
            "$and": filters
        }

    # Query
    query_args = {
        "query_texts": [question],
        "n_results": n_results
    }

    if where_filter is not None:
        query_args["where"] = where_filter

    try:

        results = collection.query(
            **query_args
        )

    except Exception as e:

        print("Filtered search error:", e)
        print("Running normal semantic search...")

        results = collection.query(
            query_texts=[question],
            n_results=n_results
        )

    return {
        "season": season,
        "team": team,
        "type": qtype,
        "results": results
    }


# ------------------------------------------------------------
# TEST QUESTIONS
# ------------------------------------------------------------

test_questions = [
    "Who won the IPL in 2016?",
    "Who won the Orange Cap in 2024?",
    "Which players were in Royal Challengers Bengaluru in 2016?",
    "How many IPL titles has Mumbai Indians won?",
    "What are Virat Kohli's IPL batting stats?"
]


for question in test_questions:

    print("\n" + "=" * 75)
    print("QUESTION:", question)

    output = smart_retrieve(
        question,
        n_results=3
    )

    print(
        "Detected →",
        "Type:", output["type"],
        "| Season:", output["season"],
        "| Team:", output["team"]
    )

    print("\nRESULTS:\n")

    for i, document in enumerate(
        output["results"]["documents"][0],
        start=1
    ):

        print(
            f"{i}. {document}"
        )

        print()
```

```python
# ============================================================
# CONNECT GEMINI TO OUR IPL RAG SYSTEM
# ============================================================

from google import genai
from getpass import getpass

# API key hidden input
GEMINI_API_KEY = getpass("Enter your Gemini API Key: ")

client = genai.Client(
    api_key=GEMINI_API_KEY
)

print("✅ Gemini client connected!")
```

```python
# ============================================================
# FINAL QUOTA-SMART HYBRID IPL AI ASSISTANT
# ============================================================

import re


# ============================================================
# DIRECT ANSWER GENERATOR
# ============================================================

def clean_direct_answer(question, retrieval):

    q = question.lower()

    qtype = retrieval["type"]

    documents = retrieval[
        "results"
    ]["documents"][0]

    if not documents:
        return None

    top_doc = documents[0]


    # ========================================================
    # SEASON SUMMARY
    # ========================================================

    if qtype == "season_summary":

        # ----------------------------------------------------
        # ORANGE CAP
        # Must be checked BEFORE general "who won"
        # ----------------------------------------------------

        if "orange cap" in q:

            match = re.search(
                r"Orange Cap winner was (.*?) with (\d+) runs",
                top_doc
            )

            if match:

                player = match.group(1)
                runs = match.group(2)

                return (
                    f"{player} won the Orange Cap "
                    f"with {runs} runs."
                )


        # ----------------------------------------------------
        # PURPLE CAP
        # ----------------------------------------------------

        if "purple cap" in q:

            match = re.search(
                r"Purple Cap winner was (.*?) with (\d+) wickets",
                top_doc
            )

            if match:

                player = match.group(1)
                wickets = match.group(2)

                return (
                    f"{player} won the Purple Cap "
                    f"with {wickets} wickets."
                )


        # ----------------------------------------------------
        # IPL CHAMPION
        # ----------------------------------------------------

        if (
            "champion" in q
            or "won the ipl" in q
            or "ipl winner" in q
            or "who won ipl" in q
        ):

            match = re.search(
                r"Champion was (.*?)\.",
                top_doc
            )

            if match:

                champion = match.group(1)

                return (
                    f"{champion} won the IPL that season."
                )


        return top_doc


    # ========================================================
    # TEAM TITLES
    # ========================================================

    if qtype == "team_titles":

        return top_doc


    # ========================================================
    # SEASON SQUAD
    # ========================================================

    if qtype == "season_squad":

        if ": " in top_doc:

            heading, players = top_doc.split(
                ": ",
                1
            )

            return (
                f"{heading}:\n\n"
                f"{players}"
            )

        return top_doc


    # ========================================================
    # CAREER BATTING
    # ========================================================

    if qtype == "career_batting":

        return top_doc


    # ========================================================
    # CAREER BOWLING
    # ========================================================

    if qtype == "career_bowling":

        return top_doc


    # ========================================================
    # POTM TOTAL
    # ========================================================

    if qtype == "potm_total":

        return top_doc


    return None


# ============================================================
# MAIN IPL AI FUNCTION
# ============================================================

def ask_ipl_ai(question):

    # ========================================================
    # 1. RETRIEVE DATA
    # ========================================================

    retrieval = smart_retrieve(
        question,
        n_results=4
    )

    documents = retrieval[
        "results"
    ]["documents"][0]


    if not documents:

        return (
            "I could not find enough information "
            "in the IPL dataset."
        )


    # ========================================================
    # 2. SIMPLE QUESTIONS → DIRECT ANSWER
    # NO GEMINI REQUIRED
    # ========================================================

    simple_types = [

        "season_summary",
        "season_squad",
        "team_titles",
        "career_batting",
        "career_bowling",
        "potm_total"

    ]


    if retrieval["type"] in simple_types:

        direct_answer = clean_direct_answer(
            question,
            retrieval
        )

        if direct_answer:

            return direct_answer


    # ========================================================
    # 3. COMPLEX QUESTIONS → GEMINI
    # ========================================================

    context = "\n".join(
        f"- {doc}"
        for doc in documents
    )


    prompt = f"""
You are IPL Analytics AI.

Answer the user's question using ONLY the IPL dataset
context given below.

Rules:

1. Never invent IPL statistics.
2. Use only the supplied IPL context.
3. Give the direct answer first.
4. Keep the answer concise but informative.
5. Player names may appear abbreviated.
6. Dataset covers IPL seasons from 2008 through 2026.
7. If the context is insufficient, clearly say that
   the dataset does not contain enough information.

IPL DATA CONTEXT:

{context}

USER QUESTION:

{question}

ANSWER:
"""


    # ========================================================
    # 4. GEMINI
    # ========================================================

    try:

        response = client.models.generate_content(
            model="gemini-3.7-flash",
            contents=prompt
        )

        if response.text:

            return response.text.strip()


    except Exception as e:

        error_text = str(e)


        # ----------------------------------------------------
        # QUOTA EXCEEDED
        # ----------------------------------------------------

        if (
            "429" in error_text
            or "RESOURCE_EXHAUSTED" in error_text
            or "quota" in error_text.lower()
        ):

            return (
                "Gemini quota is currently exhausted.\n\n"
                "Relevant IPL data:\n\n"
                + "\n".join(
                    f"• {doc}"
                    for doc in documents
                )
            )


        # ----------------------------------------------------
        # SERVER BUSY
        # ----------------------------------------------------

        if (
            "503" in error_text
            or "UNAVAILABLE" in error_text
            or "high demand" in error_text.lower()
        ):

            return (
                "Gemini is temporarily busy.\n\n"
                "Relevant IPL data:\n\n"
                + "\n".join(
                    f"• {doc}"
                    for doc in documents
                )
            )


        # ----------------------------------------------------
        # OTHER ERROR
        # ----------------------------------------------------

        return (
            "AI service error.\n\n"
            "Relevant IPL data:\n\n"
            + "\n".join(
                f"• {doc}"
                for doc in documents
            )
        )


    return "\n".join(documents)


print("✅ Final Hybrid IPL AI updated!")
```

```python
questions = [
    "Who won the Orange Cap in 2024 and how many runs did he score?",
    "Who won the Purple Cap in 2024?",
    "Who won IPL 2016?"
]

for question in questions:

    print("\n" + "=" * 70)

    print("👤 QUESTION:", question)

    print("\n🤖 IPL AI:")
    print(ask_ipl_ai(question))
```
